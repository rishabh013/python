```python
# 19. Write a Python Program to Check Armstrong Number?
# It is a number that is equal to the sum of its own digits, 
# each raised to a power equal to the number of digits in the number.

# For example, let's consider the number 153:
# It has three digits (1, 5, and 3).
# If we calculate + 1^3 + 5^3 + 3^3 we get 1 + 125 + 27, which is equal to 153.


def armstrong(num):
    size = len(str(num))
    summation = 0

    for digit in str(num):
        summation += int(digit) ** size

    if summation == num:
        return True
    else:
        return False


num = 153
print(armstrong(num))
```

```python
# 36. Write a Python Program to check if the given array is Monotonic.
# Answer: A monotonic array that moves in only one direction, non-increasing or non-decreasing, is not monotonic
# print(is_monotonic([1,2,2,3]))     True
# print(is_monotonic([3,2,1]))       True
# print(is_monotonic([1, 3, 2, 4]))  False

def is_monotonic(arr):
    increasing = True
    decreasing = True

    for i in range(len(arr) - 1):
        if arr[i] > arr[i + 1]:
            increasing = False
        if arr[i] < arr[i + 1]:
            decreasing = False

    return increasing or decreasing


# Example
print(is_monotonic([1,2,2,3]))     # True
print(is_monotonic([3,2,1]))       # True
print(is_monotonic([1, 3, 2, 4]))  # False
```

```python
# 37. Write a Python Program to Add Two Matrices.

def add_matrices(A, B):
    result = []
    for i in range(len(A)):             # rows
        row = []
        for j in range(len(A[0])):      # columns
            row.append(A[i][j] + B[i][j])
        result.append(row)

    return result


A = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]

B = [[9, 8, 7],
     [6, 5, 4],
     [3, 2, 1]]

result = add_matrices(A, B)

# Print result
for row in result:
    print(row)
```

```python
# 38 Write a Python Program to Multiply Two Matrices

def multiply_matrices(mat1, mat2):
    # Determine dimensions
    rows1 = len(mat1)
    cols1 = len(mat1[0])
    rows2 = len(mat2)
    cols2 = len(mat2[0])

    # Check if multiplication is possible
    if cols1 != rows2:
        return "Matrix multiplication is not possible."

    # Initialize result matrix
    result = [[0 for _ in range(cols2)] for _ in range(rows1)]

    # Matrix multiplication
    for i in range(rows1):
        for j in range(cols2):
            for k in range(cols1):
                result[i][j] += mat1[i][k] * mat2[k][j]

    return result


# Example matrices
matrix1 = [
    [1, 2, 3],
    [4, 5, 6]
]

matrix2 = [
    [7, 8],
    [9, 10],
    [11, 12]
]

# Multiply matrices
result_matrix = multiply_matrices(matrix1, matrix2)

# Print result
if isinstance(result_matrix, str):
    print(result_matrix)
else:
    print("Result of matrix multiplication:")
    for row in result_matrix:
        print(row)
```

```python
# 86. Please write a binary search function that searches an item in a sorted list.
#  The function should return the index of the element to be searched in the list.

def binary_search(arr, target):
    left, right = 0, len(arr) - 1

    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid                # Element found, return index
        elif arr[mid] < target:
            left = mid + 1            # Search in right half
        else:
            right = mid - 1           # Search in left half

    return -1  # Element not found

arr = [2, 5, 8, 12, 16, 23, 38]
target = 16

result = binary_search(arr, target)

if result != -1:
    print(f"Element found at index: {result}")
else:
    print(f"Element not found")
```

```python
# 87. Please write a program using a generator to print the numbers which can be divisible
#     by 5 and 7 between 0 and n in comma-separated form, while n is input by console.

def divisible_by_5_and_7(n):
    for num in range(n + 1):
        if num % 5 == 0 and num % 7 == 0:
            yield num

try:
    n = int(input("Enter a value for n: "))
    result = divisible_by_5_and_7(n)
    print(','.join(map(str, result)))   # note here map use to convert interger to string
except ValueError:
    print("Invalid input. Please enter a valid integer for n.")    
```

```python
# 98. Create a function that returns True if a given inequality expression is correct and False otherwise.

def correct_signs(str1):
    word = str1.split()

    for i in range(1, len(word), 2):
        left = int(word[i - 1])
        operator = word[i]
        right = int(word[i + 1])

        if operator == '<' and not (left < right):
            return False
        if operator == '>' and not (left > right):
            return False
            
    return True


print(correct_signs("1 < 2 < 6 < 9 > 3"))  # True
print(correct_signs("13 > 44 > 33 < 1"))    # False
```

```python
# 133 Create a function that takes a list of numbers and returns the number that's unique.

def unique(string1):
    new_dict = {}
    for x in string1:
        if x in new_dict:
            new_dict[x] = new_dict[x] +1
        else:
            new_dict[x] = 1
            
    for key, values in new_dict.items():
        if values == 1:
            return key
    
print(unique([3, 3, 3, 7, 3, 3]))
unique([0, 1, 1, 1, 1, 1, 1, 1])
```

```python
import math

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def getArea(self):
        # Calculate and return the area of the circle
        return round(math.pi * self.radius**2)

    def getPerimeter(self):
        # Calculate and return the perimeter (circumference) of the circle
        return round(2 * math.pi * self.radius)


# Test cases
circy = Circle(11)
print(circy.getArea())
print(circy.getPerimeter())
```

```python
# 135 Create a function that takes a list of strings and return a list, sorted from shortest to longest.
# sort_by_length(["Google", "Apple", "Microsoft"]) ➞ ["Apple", "Google", "Microsoft"]

def sort_by_length(list1):
    
    dict = {}
    for x in list1:
        dict[x] = len(x)
        
    new_list = []
    for y in sorted(dict.values()):
        for key, value in dict.items():
            if y == value:
                new_list.append(key)
        
    return new_list
    
sort_by_length(["Google", "Apple", "Microsoft"])
```

```python
# Write a Python function to check whether a number is perfect or not.

def perfect_number(n):
    sum_num = 0
    for x in range(1, n):
        if n % x == 0:
            sum_num = sum_num + x
    return sum_num == n
print(perfect_number(8))
```
