```python
# 1. Create the N-gram
def generate_ngrams(sentence, n):
    words = sentence.split()
    ngrams = []
    
    for i in range(len(words) - n + 1):
        ngram = tuple(words[i:i+n])
        ngrams.append(ngram)
        
    return ngrams

sentence = "I am a data scientist"

# Examples
print("Unigrams:", generate_ngrams(sentence, 1))
print("Bigrams:", generate_ngrams(sentence, 2))
print("Trigrams:", generate_ngrams(sentence, 3))
```

```python
### 2. Write the program reverse function, using buit-in function and with out buitin funtion.
def reverse_string(str1):
    new_str = ''
    for i in range(len(str1)-1, -1, -1):
        new_str = new_str + str1[i]
    return new_str

string1 = 'rishabh'
print(reverse_string(string1))


def reverse_string(str1):
    str1 = ''.join(reversed(str1))
    return str1

string1 = 'rishabh'
print(reverse_string(string1))
```

```python
# 4. Write the python code RLE compression
# Input > aaabbccccdd
# output > a3b2c4d2

def rle_compression(str1):
    dict1 = {}
    for x in str1:
        dict1[x] = dict1.get(x,0) +1
        
    for keys, items in dict1.items():
        yield f"{keys}{items}"
        
str1 = 'aaabbccccdd'

data = rle_compression(str1)
data1 = ''.join(map(str, data))
print(data1)
```

```python
# 6. Write a Python decorator with an example use case.

class Employee:
    company = "Google"   # Class variable
    def __init__(self, name):
        self.name = name

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

# Before change
print(Employee.company)

# Call class method
Employee.change_company("Microsoft")

# After change
print(Employee.company)
```

```python
#8. You are given a numpy Array and a new column as input.
# How will you delete the second column and replace the column with a new column value.
# given array = [[35,53,63], [72,22,22], [43,84,56]]
# new column value = [20,30,40]

array = [[35,53,63], [72,22,22], [43,84,56]]
new_value = [20,30,40]

new_array = []
for x in range(len(array)):
    new_list = []
    for y in range(len(array[0])):
        if y ==1:
            new_list.append(new_value[x])
        else:
            new_list.append(array[x][y])
    new_array.append(new_list)

print(new_array)

for row in new_array:
    print(row)
```

```python
# 7. What is multithreading? Write the code.

# Multithreading in Python means running multiple threads (smaller units of a process) at the same time within a single program.
# One program → multiple tasks running concurrently

from time import sleep
from threading import *

class tablet(Thread):
    def run(self):
        for i in range(5):
            print("hello")
            sleep(1)
            
class laptop(Thread):
    def run(self):
        for i in range(5):
            print("Hi")
            sleep(1)
            
t1 = tablet()
t2 = laptop()

t1.start()
sleep(0.2)
t2.start()

t1.join()
t2.join()
print("Bye")
```

```python
# What is a generator? Write the code.

# A generator is a function that uses yield.
# yield is used to generate values one at a time instead of returning everything at once.

def nums():
    for i in range(10):
        yield i

print(map(str, nums()))
print(', '.join(map(str, nums())))
```
