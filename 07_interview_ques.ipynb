```python
# 1. Create the N-gram

def generate_ngrams(sentence, n):
    words = sentence.split()
    ngrams = []
    
    for i in range(len(words) - n + 1):
        ngram = tuple(words[i:i+n])
        ngrams.append(ngram)
        
    return ngrams

sentence = "I am a data scientist"

# Examples
print("Unigrams:", generate_ngrams(sentence, 1))
print("Bigrams:", generate_ngrams(sentence, 2))
print("Trigrams:", generate_ngrams(sentence, 3))
```

```python
# 5. Write the Python code RLE compression
# Input > aaabbccccdd
# output > a3b2c4d2

def rle_compression(str1):
    dict1 = {}
    for x in str1:
        dict1[x] = dict1.get(x,0) +1
        
    for keys, items in dict1.items():
        yield f"{keys}{items}"
        
str1 = 'aaabbccccdd'

data = rle_compression(str1)
data1 = ''.join(map(str, data))
print(data1)

```

```python
def fls(str1):
    data = sorted(set(str1))
    new_str = ''
    for x in data:
        count1 = str1.count(x)
        new_str = new_str + f"{x}{count1}"
    return new_str

data = 'aaabbccccdd'

print(fls(data))
```

```python
# 8. Write a Python decorator with an example use case.

class Employee:
    company = "Google"   # Class variable
    def __init__(self, name):
        self.name = name

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

# Before change
print(Employee.company)

# Call class method
Employee.change_company("Microsoft")

# After change
print(Employee.company)
```

```python
# 11. What is multithreading? Write the code.

# Multithreading in Python means running multiple threads (smaller units of a process) at the same time within a single program.
# One program → multiple tasks running concurrently

from time import sleep
from threading import *

class tablet(Thread):
    def run(self):
        for i in range(5):
            print("hello")
            sleep(1)
            
class laptop(Thread):
    def run(self):
        for i in range(5):
            print("Hi")
            sleep(1)
            
t1 = tablet()
t2 = laptop()

t1.start()
sleep(0.2)
t2.start()

t1.join()
t2.join()
print("Bye")
```

```python
# 17. Write a Python retry decorator that retries a function multiple times if an exception occurs.

from functools import wraps

def max_attempt(max_try=3):
    def decorator(func):

        @wraps(func)
        def wrapper(*args, **kwargs):
            attempt = 0

            while attempt < max_try:
                try:
                    return func(*args, **kwargs)

                except Exception as e:
                    attempt += 1
                    print(f"Attempt {attempt} failed: {e}")

            return "Maximum attempts failed"

        return wrapper

    return decorator


@max_attempt(max_try=3)
def call_llm(prompt: str) -> str:

    # Simulating LLM call
    if prompt == "error":
        raise Exception("LLM API failed")

    return f"answer: {prompt}"


print(call_llm("hello"))
print(call_llm("error"))
```

```python
# 19. Reverse only the vowels in a string while keeping all other characters in their original positions
# Input = "IceCreAm"
# output = "AceCreIm"

input_str = "IceCreAm"
vowels = "aeiou"

# Step 1: Extract all vowels
vowel_chars = ""

for ch in input_str:
    if ch.lower() in vowels:
        vowel_chars += ch

# Step 2: Reverse the extracted vowels
vowel_chars = vowel_chars[::-1]

print("Reversed vowels:", vowel_chars)

# Step 3: Replace original vowels with reversed vowels
result = ""
index = 0

for ch in input_str:
    if ch.lower() in vowels:
        result += vowel_chars[index]
        index += 1
    else:
        result += ch

print("Output:", result)
```
